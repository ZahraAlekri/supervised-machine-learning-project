# Loan Approval Outcome Prediction
## Supervised Binary Classification — End-to-End Machine Learning Pipeline

**Dataset:** Loan Approval Dataset (47,250 applicants)  
**Target Variable:** `loan_status` — 1 = Approved, 0 = Rejected  
**Task:** Predict whether a loan application will be approved based on demographic, financial, and credit history features.

---


## 1. Imports & Setup

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, ConfusionMatrixDisplay,
    classification_report
)

# Consistent plot style
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 120
```

## 2. Load Data

We load the dataset and perform an initial inspection to understand its structure, data types, and any obvious issues before diving into cleaning.


```
df = pd.read_csv('loan_data.csv')

# Drop the unnamed index column — it carries no information
df.drop(columns=['Unnamed: 0'], inplace=True)

print(f"Dataset shape: {df.shape}")
df.head()
```

## 4. Data Cleaning & Preprocessing

Based on the EDA findings, we apply the following cleaning steps:

| Issue | Action |
|---|---|
| `Unnamed: 0` index column | Drop |
| `$` signs in `person_income` and `loan_amnt` | Strip and cast to float |
| Inconsistent casing & whitespace in categoricals | Strip + lowercase + standardise |
| `person_age` > 80 (impossible values) | Remove rows |
| `person_emp_exp` > 60 (impossible values) | Remove rows |

We **do not** drop `person_gender` despite it being a potential fairness concern — we keep it for analysis and let the model decide its predictive value.


```
df_clean = df.copy()

# ── 4.1 Parse money columns ──────────────────────────────────────────────────
def parse_money(series):
    """Remove $ sign from either end and cast to float."""
    return series.astype(str).str.replace('$', '', regex=False).str.strip().astype(float)

df_clean['person_income'] = parse_money(df_clean['person_income'])
df_clean['loan_amnt']     = parse_money(df_clean['loan_amnt'])

# ── 4.2 Standardise categorical strings ──────────────────────────────────────
str_cols = ['person_gender', 'person_education', 'person_home_ownership',
            'loan_intent', 'previous_loan_defaults_on_file']

for col in str_cols:
    df_clean[col] = df_clean[col].str.strip().str.lower()

# Manual consolidation for education (mixed case artefacts)
edu_map = {
    'master':       'master',
    'high school':  'high school',
    'bachelor':     'bachelor',
    'associate':    'associate',
    'doctorate':    'doctorate',
}
df_clean['person_education'] = df_clean['person_education'].map(edu_map)

# ── 4.3 Remove impossible outlier rows ───────────────────────────────────────
before = len(df_clean)
df_clean = df_clean[df_clean['person_age'] <= 80]
df_clean = df_clean[df_clean['person_emp_exp'] <= 60]
after = len(df_clean)
print(f"Rows removed (outliers): {before - after}  |  Remaining: {after}")

# ── 4.4 Verify ───────────────────────────────────────────────────────────────
print()
print("Remaining nulls:")
print(df_clean.isnull().sum())
df_clean.dtypes
```

## 5. Feature Engineering

We create new features that capture relationships not directly expressed in the raw columns. Each feature is motivated by domain logic.

| New Feature | Formula | Rationale |
|---|---|---|
| `debt_to_income` | `loan_amnt / person_income` | Higher ratio = higher repayment burden |
| `income_per_exp_year` | `person_income / (person_emp_exp + 1)` | Normalises income by career length; +1 avoids division by zero |
| `education_level` | Ordinal encoding of education | Captures educational progression numerically |
| `has_default` | Binary from `previous_loan_defaults_on_file` | Simplifies the yes/no into 0/1 |


```
df_feat = df_clean.copy()

# ── 5.1 Debt-to-income ratio ─────────────────────────────────────────────────
df_feat['debt_to_income'] = df_feat['loan_amnt'] / df_feat['person_income']

# ── 5.2 Income per year of experience ────────────────────────────────────────
df_feat['income_per_exp_year'] = df_feat['person_income'] / (df_feat['person_emp_exp'] + 1)

# ── 5.3 Ordinal encoding for education ───────────────────────────────────────
edu_order = {
    'high school': 0,
    'associate':   1,
    'bachelor':    2,
    'master':      3,
    'doctorate':   4
}
df_feat['education_level'] = df_feat['person_education'].map(edu_order)

# ── 5.4 Binary default flag ──────────────────────────────────────────────────
df_feat['has_default'] = (df_feat['previous_loan_defaults_on_file'] == 'yes').astype(int)

# ── 5.5 One-hot encode remaining categoricals ────────────────────────────────
ohe_cols = ['person_gender', 'person_home_ownership', 'loan_intent']
df_feat = pd.get_dummies(df_feat, columns=ohe_cols, drop_first=True)

# ── 5.6 Drop original columns now replaced ───────────────────────────────────
df_feat.drop(columns=['person_education', 'previous_loan_defaults_on_file'], inplace=True)

print(f"Final feature set shape: {df_feat.shape}")
print()
df_feat.head()
```

## 6. Train / Test Split

We use a **stratified 80/20 split** to preserve the class ratio in both sets. This is important given the class imbalance (~78% rejected).

Stratified splitting ensures the model sees a representative proportion of approvals and rejections during both training and evaluation.


```
X = df_feat.drop(columns=['loan_status'])
y = df_feat['loan_status']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y          # preserves class balance in both splits
)

print(f"Training set:  {X_train.shape[0]} rows")
print(f"Test set:      {X_test.shape[0]} rows")
print()
print("Training class balance:")
print(y_train.value_counts(normalize=True).round(3))
print()
print("Test class balance:")
print(y_test.value_counts(normalize=True).round(3))
```

## 7. Evaluation Helper Function

We define a reusable function that prints all required metrics and plots the confusion matrix. This keeps our evaluation consistent and compact across all models.


```
def evaluate_model(name, model, X_tr, X_te, y_tr, y_te):
    """
    Fit the model, print classification metrics, and plot the confusion matrix.
    Returns a dict of metrics for the final comparison table.
    """
    model.fit(X_tr, y_tr)
    y_pred = model.predict(X_te)

    acc  = accuracy_score(y_te, y_pred)
    prec = precision_score(y_te, y_pred)
    rec  = recall_score(y_te, y_pred)
    f1   = f1_score(y_te, y_pred)

    print(f"\n{'='*55}")
    print(f"  Model: {name}")
    print(f"{'='*55}")
    print(f"  Accuracy  : {acc:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}")
    print(f"  F1-Score  : {f1:.4f}")
    print()
    print(classification_report(y_te, y_pred, target_names=['Rejected', 'Approved']))

    # Confusion matrix
    cm = confusion_matrix(y_te, y_pred)
    disp = ConfusionMatrixDisplay(cm, display_labels=['Rejected', 'Approved'])
    fig, ax = plt.subplots(figsize=(5, 4))
    disp.plot(ax=ax, colorbar=False, cmap='Blues')
    ax.set_title(f'Confusion Matrix — {name}', fontsize=12)
    plt.tight_layout()
    plt.show()

    return {'Model': name, 'Accuracy': acc, 'Precision': prec,
            'Recall': rec, 'F1-Score': f1}

results = []  # collect all model results here
```

## 8. Baseline Models

We build two baseline classifiers without any hyperparameter tuning. These serve as reference benchmarks.

> **Note on scaling:** Both Logistic Regression and KNN are sensitive to feature magnitude. We scale features using `StandardScaler` (zero mean, unit variance) directly within each pipeline to prevent data leakage from the test set.

### 8.1 Baseline Logistic Regression


```
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

# Baseline LR — default hyperparameters, no regularisation tuning
lr_base = LogisticRegression(random_state=42, max_iter=1000)

res = evaluate_model(
    'Baseline Logistic Regression',
    lr_base,
    X_train_scaled, X_test_scaled,
    y_train, y_test
)
results.append(res)
```

**Interpretation:**
- Logistic Regression models the log-odds of approval as a linear combination of features.
- Default regularisation (`C=1.0`, L2) is applied by scikit-learn unless changed.
- Given the class imbalance, pay close attention to Recall for the approved class — falsely rejecting a creditworthy applicant is a real business cost.

### 8.2 Baseline KNN


```
# Baseline KNN — default k=5, no tuning
knn_base = KNeighborsClassifier(n_neighbors=5)

res = evaluate_model(
    'Baseline KNN (k=5)',
    knn_base,
    X_train_scaled, X_test_scaled,
    y_train, y_test
)
results.append(res)
```

**Interpretation:**
- KNN classifies a sample by majority vote among its k nearest neighbours in feature space.
- It is non-parametric — it makes no assumptions about the underlying distribution.
- Very sensitive to feature scale (handled by our StandardScaler) and to the choice of k.


## 9. Hyperparameter Tuning with GridSearchCV & Pipelines

We now improve both models using:
- **Pipeline**: chains scaling and the classifier into a single object, preventing data leakage and streamlining cross-validation.
- **GridSearchCV**: exhaustively searches over specified hyperparameter combinations using cross-validation.
- **Regularisation** (Logistic Regression): we search over `C` (inverse regularisation strength) and penalty type.

We use `StratifiedKFold(n_splits=5)` to respect class imbalance during CV, and optimise for **F1-Score** since we care about both precision and recall on an imbalanced dataset.

### 9.1 Tuned Logistic Regression


```
# Pipeline: scale → logistic regression
lr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(random_state=42, max_iter=1000))
])

# Hyperparameter grid
lr_param_grid = {
    'clf__C':       [0.01, 0.1, 1, 10, 100],   # regularisation strength
    'clf__penalty': ['l1', 'l2'],               # regularisation type
    'clf__solver':  ['liblinear']               # supports both l1 and l2
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

lr_grid = GridSearchCV(
    lr_pipeline,
    lr_param_grid,
    cv=cv,
    scoring='f1',
    n_jobs=-1,
    verbose=0
)

lr_grid.fit(X_train, y_train)

print(f"Best LR hyperparameters: {lr_grid.best_params_}")
print(f"Best CV F1-Score:        {lr_grid.best_score_:.4f}")
```

```
res = evaluate_model(
    'Tuned Logistic Regression',
    lr_grid.best_estimator_,
    X_train, X_test,
    y_train, y_test
)
results.append(res)
```

**What changed:** GridSearchCV selected the optimal combination of `C` (regularisation strength) and penalty type. A smaller `C` applies stronger regularisation, reducing overfitting; a larger `C` trusts the data more and fits a less constrained boundary.

### 9.2 Tuned KNN


```
# Pipeline: scale → KNN
knn_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', KNeighborsClassifier())
])

# Hyperparameter grid
knn_param_grid = {
    'clf__n_neighbors': [3, 5, 9, 15, 21, 31],  # number of neighbours
    'clf__weights':     ['uniform', 'distance'],  # equal vs distance-weighted vote
    'clf__metric':      ['euclidean', 'manhattan']  # distance measure
}

knn_grid = GridSearchCV(
    knn_pipeline,
    knn_param_grid,
    cv=cv,
    scoring='f1',
    n_jobs=-1,
    verbose=0
)

knn_grid.fit(X_train, y_train)

print(f"Best KNN hyperparameters: {knn_grid.best_params_}")
print(f"Best CV F1-Score:         {knn_grid.best_score_:.4f}")
```

```
res = evaluate_model(
    'Tuned KNN',
    knn_grid.best_estimator_,
    X_train, X_test,
    y_train, y_test
)
results.append(res)
```

## 10. Comparative Model Analysis

We now bring all four models side by side to compare their performance across all metrics.


```
results_df = pd.DataFrame(results).set_index('Model')
results_df = results_df.round(4)
results_df
```

## 11. Conclusion & Recommendations

### Which model performed best?

After comparing all four models across Accuracy, Precision, Recall, and F1-Score, the **Tuned Logistic Regression** emerges as the preferred model for this task. Here is why:

- **Interpretability:** Logistic Regression produces coefficients that directly indicate each feature's contribution to the approval decision. This matters in a lending context where regulators may require explanations.
- **Competitive performance:** The tuned version, with the optimal regularisation strength selected via GridSearchCV, achieves a strong balance between Precision and Recall — reflected in its F1-Score.
- **Generalisation:** Regularisation (L1 or L2) prevents the model from overfitting noisy patterns in the training data, leading to more reliable predictions on new applicants.

KNN, while non-parametric and flexible, is computationally expensive at inference time (it must compare against all training points) and its performance is less stable across different neighbourhood configurations.

### Key findings from EDA:
- **Previous loan defaults** is the single strongest predictor of rejection — applicants with a default history are overwhelmingly rejected.
- **Credit score** and **loan-to-income ratio** also show strong separation between approved and rejected applicants.
- **Education level** showed moderate influence — higher education correlates with slightly higher approval rates.
- The dataset is significantly **imbalanced** (78% rejected). Optimising for accuracy alone would be misleading — a model that rejects everyone would score ~78% accuracy while being commercially useless.

### Business Recommendations:
1. **Do not rely on accuracy alone** when evaluating lending models. Use F1-Score or the Precision-Recall tradeoff as the primary metric, depending on whether the business cost of false approvals or false rejections is higher.
2. **Applicants with prior defaults** should be flagged for additional manual review rather than automatic rejection — there may be contextual factors (e.g., time since default, recovery of credit score) that warrant exceptions.
3. **Debt-to-income ratio** is a reliable engineered signal. Incorporating it into underwriting rules could improve consistency across loan officers.
4. Consider collecting additional data (e.g., payment history, outstanding debts) to further improve model performance before deployment.
